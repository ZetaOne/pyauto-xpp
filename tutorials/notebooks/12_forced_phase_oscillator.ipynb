# Forced phase oscillator on a torus: Arnold tongues

`x` is the phase of an oscillator forced at frequency `om`; `y` is the phase of the
forcing. Both are angles (a torus). For each forcing amplitude `b` there are ranges of
`om` where the oscillator locks to the forcing (Arnold tongues).

*Based on an example from XPP Studio (Peter Ashwin and Bard Ermentrout, GPL-3.0). Every number below is computed by XPPAUT 8.1 and its AUTO through pyauto-xpp.*

```python
import plotly.graph_objects as go
from pyxpp.builder import ModelBuilder
from xpptut import Explorer, load, orbit_diagram, show_figure
```

## The model

Written with `ModelBuilder`, statement by statement as in an `.ode` file. Edit it and rerun the cells below.

```python
mb = (ModelBuilder("forced_phase_oscillator")
      .function("p(x)", "-sin(x)/(1+a*cos(x))")
      .ode("x", "1+b*p(x)*cos(y)")
      .ode("y", "om")
      .parameter(om=2, b=0.25, a=0.25)
      .initial(x=0, y=0)
      .option(fold="x", tor_per=6.283185)
      .option(fold="y")
      .option(total=100, dt=0.05))
print(mb.render())
```

## Explore

The sliders, the direction-field menu and the AUTO panel all call XPP. `ex.m` is the pyxpp model, for anything else.

```python
ex = Explorer(mb, kind="plane", x="x", y="y", window=(0, 6.3, 0, 6.3), df=False, nullclines=False,
              fold={"x": 6.283185, "y": 6.283185}, sliders={"om": (0.5, 3.2), "b": (0, 0.5)},
              title="Forced phase oscillator")
ex
```

## Going further

**Rotation number** (mean dx/dt per forcing cycle, relative to `om`) against `om`: the
flat steps are the locked ranges.

```python
import numpy as np
from pyxpp.builder import ModelBuilder
unfolded = (ModelBuilder("phase_unfolded").function("p(x)", "-sin(x)/(1+a*cos(x))")
            .ode("x", "1+b*p(x)*cos(y)").ode("y", "om").parameter(om=2, b=0.25, a=0.25)
            .option(total=600, dt=0.05, bound=1e9))
unfolded = load(unfolded)
OM = np.linspace(0.5, 3.2, 120)
rho = []
for o in OM:
    tr = unfolded.run(keep=False, om=o, b=0.4, x=0, y=0, total=600, trans=100)
    rho.append((tr["x"][-1] - tr["x"][0]) / (tr["y"][-1] - tr["y"][0]))
show_figure(go.Figure(go.Scatter(x=OM, y=rho, mode="lines")), xaxis_title="om", yaxis_title="rotation number", height=320)
```
