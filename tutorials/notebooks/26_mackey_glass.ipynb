# Mackey-Glass: delay-induced chaos

Production depends on the population τ time units ago. Periodic for small τ,
period-doubling and chaos as τ grows (chaotic at τ = 17).

*Based on an example from XPP Studio (Peter Ashwin and Bard Ermentrout, GPL-3.0). Every number below is computed by XPPAUT 8.1 and its AUTO through pyauto-xpp.*

```python
import plotly.graph_objects as go
from pyxpp.builder import ModelBuilder
from xpptut import Explorer, load, orbit_diagram, show_figure
```

## The model

Written with `ModelBuilder`, statement by statement as in an `.ode` file. Edit it and rerun the cells below.

```python
mb = (ModelBuilder("mackey_glass")
      .ode("x", "b*delay(x,tau)/(1+delay(x,tau)^pw)-g*x")
      .aux("xlag", "delay(x,tau)")
      .parameter(b=0.2, g=0.1, pw=10, tau=17)
      .initial(x=0.5)
      .option(total=1000, dt=0.1, trans=300, delay=40))
print(mb.render())
```

## Explore

The sliders, the direction-field menu and the AUTO panel all call XPP. `ex.m` is the pyxpp model, for anything else.

```python
ex = Explorer(mb, kind="series", x="xlag", y="x", window=(0.2, 1.5, 0.2, 1.5), sliders={"tau": (2, 30)},
              title="Mackey-Glass")
ex
```

## Going further

**Maxima against τ.**

```python
import numpy as np
T_, M = [], []
for tau in np.linspace(4, 25, 60):
    x = ex.m.run(keep=False, tau=tau, total=1500, trans=800)["x"]
    k = np.where((x[1:-1] > x[:-2]) & (x[1:-1] > x[2:]))[0] + 1
    T_ += [tau] * len(k); M += list(x[k])
show_figure(go.Figure(go.Scattergl(x=T_, y=M, mode="markers", marker=dict(size=2, color="black"))),
            xaxis_title="tau", yaxis_title="maxima of x", height=360)
```
