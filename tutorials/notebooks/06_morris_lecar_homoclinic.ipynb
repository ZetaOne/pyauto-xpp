# Morris-Lecar with a saddle: a homoclinic orbit

With `gca = 4` and `phi = 0.23` the model has a saddle, and the periodic orbit born at
the Hopf point grows until it hits the saddle: its period grows without bound near
`i = 35.007` (a homoclinic bifurcation).

*Based on an example from XPP Studio (Peter Ashwin and Bard Ermentrout, GPL-3.0). Every number below is computed by XPPAUT 8.1 and its AUTO through pyauto-xpp.*

```python
import plotly.graph_objects as go
from pyxpp.builder import ModelBuilder
from xpptut import Explorer, load, orbit_diagram, show_figure
```

## The model

Written with `ModelBuilder`, statement by statement as in an `.ode` file. Edit it and rerun the cells below.

```python
mb = (ModelBuilder("morris_lecar_homoclinic")
      .function("minf(v)", ".5*(1+tanh((v-v1)/v2))")
      .function("winf(v)", ".5*(1+tanh((v-v3)/v4))")
      .function("tauw(v)", "1/cosh((v-v3)/(2*v4))")
      .ode("v", "(i+gl*(vl-v)+gk*w*(vk-v)+gca*minf(v)*(vca-v))/c")
      .ode("w", "phi*(winf(v)-w)/tauw(v)")
      .parameter(i=37, c=20, vk=-84, vl=-60, vca=120, gk=8, gl=2, gca=4)
      .parameter(v1=-1.2, v2=18, v3=12, v4=17.4, phi=0.23)
      .initial(v=-60, w=0.015)
      .option(total=400, dt=0.25, bound=10000))
print(mb.render())
```

## Explore

The sliders, the direction-field menu and the AUTO panel all call XPP. `ex.m` is the pyxpp model, for anything else.

```python
ex = Explorer(mb, kind="plane", x="v", y="w", window=(-75, 45, -0.1, 0.55),
              sliders={"i": (0, 60), "phi": (0.05, 0.6)},
              auto="i", auto_range=(0, 60), title="Morris-Lecar homoclinic")
ex
```

## Going further

**The period blows up.** Continue the orbits from the Hopf point and plot their period:
it grows like log(1/|i - i_hom|) towards the homoclinic point.

```python
parts = ex.run_auto()
hb = [p for d in parts for p in d.special("HB")]
print("Hopf points:", [round(p.x, 4) for p in hb])
ex.pick("HB", max(hb, key=lambda p: p.x).x)
po, l1 = ex.periodic()
pts = [p for p in po if p.period < 1e4]
show_figure(go.Figure(go.Scatter(x=[p.x for p in pts], y=[p.period for p in pts], mode="markers")),
            xaxis_title="i", yaxis_title="period", yaxis_type="log", height=320)
print("largest period", round(max(p.period for p in pts), 1), "at i =", round(max(pts, key=lambda p: p.period).x, 4))
```
