```python
import plotly.graph_objects as go
from pyxpp.builder import ModelBuilder
from xpptut import Explorer, load, orbit_diagram, show_figure
```

## The model

Written with `ModelBuilder`, statement by statement as in an `.ode` file. Edit it and rerun the cells below.

```python
mb = (ModelBuilder("rossler")
      .ode("x", "-y-z")
      .ode("y", "x+a*y")
      .ode("z", "b+z*(x-c)")
      .parameter(a=0.2, b=0.2, c=5.7)
      .initial(x=1, y=1, z=0)
      .option(total=300, dt=0.02, meth="rk4", bound=1000))
print(mb.render())
```

## Explore

The sliders, the direction-field menu and the AUTO panel all call XPP. `ex.m` is the pyxpp model, for anything else.

```python
ex = Explorer(mb, kind="3d", x="x", y="y", z="z", sliders={"c": (2, 6.5), "a": (0.05, 0.4)}, title="Rössler")
ex
```

## Going further

**Maxima of x against c**: one maximum per cycle, two after the first doubling, ...

```python
import numpy as np
C, M = [], []
for c in np.linspace(2, 6.5, 160):
    tr = ex.m.run(keep=False, c=c, total=500, trans=300)
    x = tr["x"]; k = np.where((x[1:-1] > x[:-2]) & (x[1:-1] > x[2:]))[0] + 1
    C += [c] * len(k); M += list(x[k])
show_figure(go.Figure(go.Scattergl(x=C, y=M, mode="markers", marker=dict(size=2, color="black"))),
            xaxis_title="c", yaxis_title="maxima of x", height=380)
```
