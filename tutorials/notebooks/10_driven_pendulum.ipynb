# Damped driven pendulum on a cylinder

`th` is an angle, so the phase space is a cylinder: XPP folds `th` into [0, 2π).
With torque `i` above a critical value the pendulum rotates forever; below it, it may
rest or rotate depending on where it starts.

*Based on an example from XPP Studio (Peter Ashwin and Bard Ermentrout, GPL-3.0). Every number below is computed by XPPAUT 8.1 and its AUTO through pyauto-xpp.*

```python
import plotly.graph_objects as go
from pyxpp.builder import ModelBuilder
from xpptut import Explorer, load, orbit_diagram, show_figure
```

## The model

Written with `ModelBuilder`, statement by statement as in an `.ode` file. Edit it and rerun the cells below.

```python
mb = (ModelBuilder("driven_pendulum")
      .ode("th", "v")
      .ode("v", "i-b*v-sin(th)")
      .parameter(i=0.5, b=0.3)
      .initial(th=0, v=2.5)
      .option(fold="th", tor_per=6.283185307, total=100, dt=0.05))
print(mb.render())
```

## Explore

The sliders, the direction-field menu and the AUTO panel all call XPP. `ex.m` is the pyxpp model, for anything else.

```python
ex = Explorer(mb, kind="plane", x="th", y="v", window=(0, 6.2832, -2, 4), fold={"th": 6.283185307},
              sliders={"i": (0, 1.5), "b": (0.05, 1)}, title="Driven pendulum")
ex
```

## Going further

**Rest or rotate?** From the same start, the long-time mean velocity against torque.

```python
import numpy as np
I = np.linspace(0, 1.5, 31)
mean_v = [ex.m.run(keep=False, i=x, th=0, v=2.5, total=400, trans=200)["v"].mean() for x in I]
show_figure(go.Figure(go.Scatter(x=I, y=mean_v, mode="lines+markers")), xaxis_title="i",
            yaxis_title="mean velocity", height=300)
```
