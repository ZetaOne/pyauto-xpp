# Integrate-and-fire neuron with adaptation

`v` charges towards `i`; when it reaches the threshold 1 a `global` event resets it to 0
and kicks the adaptation `w` by `b`. Adaptation makes the inter-spike intervals grow
until they settle.

*Based on an example from XPP Studio (Peter Ashwin and Bard Ermentrout, GPL-3.0). Every number below is computed by XPPAUT 8.1 and its AUTO through pyauto-xpp.*

```python
import plotly.graph_objects as go
from pyxpp.builder import ModelBuilder
from xpptut import Explorer, load, orbit_diagram, show_figure
```

## The model

Written with `ModelBuilder`, statement by statement as in an `.ode` file. Edit it and rerun the cells below.

```python
mb = (ModelBuilder("integrate_and_fire")
      .ode("v", "(-v+i-w)/tau")
      .ode("w", "-w/tauw")
      .global_event(1, "v-1", {"v": 0, "w": "w+b"})
      .parameter(i=1.6, tau=1, tauw=20, b=0.2)
      .initial(v=0, w=0)
      .option(total=100, dt=0.01))
print(mb.render())
```

## Explore

The sliders, the direction-field menu and the AUTO panel all call XPP. `ex.m` is the pyxpp model, for anything else.

```python
ex = Explorer(mb, kind="plane", x="v", y="w", window=(-0.1, 1.1, -0.05, 1), nullclines=False,
              sliders={"i": (0.5, 3), "b": (0, 1)}, title="Integrate-and-fire")
ex
```

## Going further

**Inter-spike intervals**: they lengthen as `w` builds up, then settle.

```python
import numpy as np
tr = ex.m.run(keep=False, total=200)
spikes = tr.t[1:][np.diff(tr["v"]) < -0.5]
isi = np.diff(spikes)
show_figure(go.Figure(go.Scatter(y=isi, mode="lines+markers")), xaxis_title="spike number", yaxis_title="interval", height=300)
```
