```python
import plotly.graph_objects as go
from pyxpp.builder import ModelBuilder
from xpptut import Explorer, load, orbit_diagram, show_figure
```

## The model

Written with `ModelBuilder`, statement by statement as in an `.ode` file. Edit it and rerun the cells below.

```python
mb = (ModelBuilder("henon_map")
      .map("x", "1-a*x^2+y")
      .map("y", "b*x")
      .parameter(a=1.4, b=0.3)
      .initial(x=0, y=0)
      .option(total=10000, meth="discrete", bound=100))
print(mb.render())
```

## Explore

The sliders, the direction-field menu and the AUTO panel all call XPP. `ex.m` is the pyxpp model, for anything else.

```python
ex = Explorer(mb, kind="map", x="x", y="y", window=(-1.6, 1.6, -0.5, 0.5), sliders={"a": (-0.2, 1.45), "b": (-0.4, 0.4)},
              title="Hénon map")
ex
```

## Going further

**Orbit diagram in `a`** (b = 0.3).

```python
import numpy as np
P, X = orbit_diagram(mb, "a", np.linspace(0, 1.42, 500), "x", transient=400, keep=120)
show_figure(go.Figure(go.Scattergl(x=P, y=X, mode="markers", marker=dict(size=1, color="black"))),
            xaxis_title="a", yaxis_title="x", height=380)
```

**Largest Liapunov exponent** from XPP (stocHast > Liapunov).

```python
m = load(mb); m.run()
m.buttons.numerics.stochast.liapunov(0)
print(m.transcript.strip().splitlines()[-1])
```
