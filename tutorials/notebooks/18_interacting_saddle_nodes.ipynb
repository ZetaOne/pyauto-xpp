# Interacting saddle-nodes

From Ashwin, Postlethwaite and Sieber (Nonlinearity 2026): two saddle-node
bifurcations that interact through the coupling `alpha`, `beta`; `lambda` is built from
`mu` and `gamma`. The file's initial point (1, 1) leaves the box at once (XPP stops at
`bound = 10`), so start trajectories by clicking near the equilibria.

*Based on an example from XPP Studio (Peter Ashwin and Bard Ermentrout, GPL-3.0). Every number below is computed by XPPAUT 8.1 and its AUTO through pyauto-xpp.*

```python
import plotly.graph_objects as go
from pyxpp.builder import ModelBuilder
from xpptut import Explorer, load, orbit_diagram, show_figure
```

## The model

Written with `ModelBuilder`, statement by statement as in an `.ode` file. Edit it and rerun the cells below.

```python
mb = (ModelBuilder("interacting_saddle_nodes")
      .parameter(mu=-0.005, gamma=3.58)
      .parameter(alpha=3.1, beta=-1.3)
      .fixed("lambda", "mu + gamma*(gamma+alpha^2-beta^2)/(alpha+beta)^2")
      .ode("x", "x^2-lambda+2*alpha*y-gamma")
      .ode("y", "y^2-lambda+2*beta*x+gamma")
      .initial(x=1, y=1)
      .option(maxstor=10000, epsu=0.000001, epss=0.000001)
      .option(meth="dopri5", total=10, dt=0.01, toler=0.000001, bound=10)
      .option(ds=0.01, dsmin=0.0001, dsmax=0.05, nmax=3000, parmin=-1, parmax=1))
print(mb.render())
```

## Explore

The sliders, the direction-field menu and the AUTO panel all call XPP. `ex.m` is the pyxpp model, for anything else.

```python
ex = Explorer(mb, kind="plane", x="x", y="y", window=(-8.5, 19.6, -13.2, 5.9),
              sliders={"mu": (-1, 1), "gamma": (2, 5)},
              auto="mu", auto_range=(-1, 1), auto2="gamma", auto2_range=(2, 5), title="Interacting saddle-nodes")
ex
```

## Going further

**The equilibria and the branch in `mu`.** AUTO follows the equilibrium the explorer
starts from; on this branch there is no fold or Hopf point between mu = -1 and 1.

```python
print([(e["kind"], round(e["x"], 3), round(e["y"], 3)) for e in ex.eqs])
parts = ex.run_auto()
special = sorted({(p.type, round(p.x, 5)) for d in parts for p in d.labeled() if p.type in ("LP", "HB", "BP")})
print(special or "no fold or Hopf point on this branch for mu in [-1, 1]")
```
