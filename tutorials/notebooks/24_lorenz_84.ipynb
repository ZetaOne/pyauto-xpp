# Lorenz 84: a model of the atmosphere

Lorenz's 1984 low-order model of the general circulation: `x` is the westerly wind,
`y`, `z` a chain of travelling eddies; `f` and `g` are thermal forcings. F = 8 gives chaos;
lower F gives periodic flow, and the steady westerly flow loses stability in a Hopf
bifurcation.

*Based on an example from XPP Studio (Peter Ashwin and Bard Ermentrout, GPL-3.0). Every number below is computed by XPPAUT 8.1 and its AUTO through pyauto-xpp.*

```python
import plotly.graph_objects as go
from pyxpp.builder import ModelBuilder
from xpptut import Explorer, load, orbit_diagram, show_figure
```

## The model

Written with `ModelBuilder`, statement by statement as in an `.ode` file. Edit it and rerun the cells below.

```python
mb = (ModelBuilder("lorenz84")
      .ode("x", "-y^2-z^2-a*x+a*f")
      .ode("y", "x*y-b*x*z-y+g")
      .ode("z", "b*x*y+x*z-z")
      .parameter(a=0.25, b=4, f=8, g=1)
      .initial(x=1, y=0, z=0)
      .option(total=200, dt=0.02, trans=20))
print(mb.render())
```

## Explore

The sliders, the direction-field menu and the AUTO panel all call XPP. `ex.m` is the pyxpp model, for anything else.

```python
ex = Explorer(mb, kind="3d", x="y", y="z", z="x", sliders={"f": (0, 10), "g": (0, 2)}, title="Lorenz 84")
ex
```

## Going further

**The steady westerly flow and its Hopf point in `f`.**

```python
m = load(mb)
m.set(f=1); m.run(total=500); m.run_last(total=10)
d = m.auto.run("f", start="steady", ds=0.01, dsmax=0.05, parmin=0, parmax=10, nmax=1000)
print([(p.type, round(p.x, 4)) for p in d.labeled() if p.type in ("HB", "LP")])
```
