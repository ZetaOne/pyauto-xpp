An area-preserving twist map on the torus. For large `K` most orbits wander
chaotically; for smaller `K` invariant curves and islands survive. Click several starts
to see different orbits.

```python
import plotly.graph_objects as go
from pyxpp.builder import ModelBuilder
from xpptut import Explorer, load, orbit_diagram, show_figure
```

## The model

Written with `ModelBuilder`, statement by statement as in an `.ode` file. Edit it and rerun the cells below.

```python
mb = (ModelBuilder("chirikov_taylor_map")
      .fixed("temp", "p+K*sin(q)/6.28318530718")
      .map("p", "temp")
      .map("q", "q+temp")
      .parameter(K=6.0)
      .initial(p=4.1, q=1.2)
      .option(total=3000, meth="discrete", bound=100)
      .option(fold="p", tor_per=6.28318530718)
      .option(fold="q"))
print(mb.render())
```

## Explore

The sliders, the direction-field menu and the AUTO panel all call XPP. `ex.m` is the pyxpp model, for anything else.

```python
ex = Explorer(mb, kind="map", x="q", y="p", window=(0, 6.2832, 0, 6.2832), sliders={"K": (0, 8)},
              title="Chirikov-Taylor map")
ex
```

## Going further

**A phase portrait from many starts** at K = 0.9 (islands and invariant curves).

```python
import numpy as np
m = load(mb)
fig = go.Figure()
for k, (q0, p0) in enumerate(zip(np.linspace(0.1, 6.2, 18), np.linspace(0.2, 6.1, 18))):
    tr = m.run(keep=False, K=0.9, q=q0, p=p0, total=800)
    fig.add_scattergl(x=tr["q"], y=tr["p"], mode="markers", marker=dict(size=1.5), showlegend=False)
show_figure(fig, xaxis_title="q", yaxis_title="p", height=420)
```
