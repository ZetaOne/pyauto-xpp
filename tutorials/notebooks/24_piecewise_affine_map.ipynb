From Ashwin & Goetz (SIADS 2006): y → frac(−x + a y). XPP Studio's file uses `floor`;
XPP's own name is `flr`, used here.

```python
import plotly.graph_objects as go
from pyxpp.builder import ModelBuilder
from xpptut import Explorer, load, orbit_diagram, show_figure
```

## The model

Written with `ModelBuilder`, statement by statement as in an `.ode` file. Edit it and rerun the cells below.

```python
mb = (ModelBuilder("piecewise_affine_map")
      .parameter(a=0.5)
      .fixed("yy", "-x+a*y")
      .fixed("ny", "yy-flr(yy)")
      .map("x", "y")
      .map("y", "ny")
      .initial(x=0.4319130445, y=0.2635796134)
      .option(maxstor=40000, meth="discrete", total=30000, bound=10000))
print(mb.render())
```

## Explore

The sliders, the direction-field menu and the AUTO panel all call XPP. `ex.m` is the pyxpp model, for anything else.

```python
ex = Explorer(mb, kind="map", x="x", y="y", window=(-0.08, 1.07, -0.08, 1.07), sliders={"a": (0, 1)},
              title="Piecewise affine map")
ex
```

## Going further

**Several values of `a` side by side.**

```python
from plotly.subplots import make_subplots
m = load(mb)
fig = make_subplots(rows=1, cols=3, subplot_titles=["a = 0.5", "a = 0.618", "a = 0.9"])
for k, a in enumerate((0.5, 0.618, 0.9)):
    tr = m.run(keep=False, a=a, total=20000)
    fig.add_scattergl(x=tr["x"], y=tr["y"], mode="markers", marker=dict(size=1), row=1, col=k + 1, showlegend=False)
show_figure(fig, height=360)
```
