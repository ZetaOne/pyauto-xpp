# Hutchinson's delayed logistic equation

x' = r x (1 − x(t − τ)). The equilibrium x = 1 loses stability in a Hopf bifurcation
at r τ = π/2. The history for t < 0 is given by a formula; `xlag` is the delayed value.

*Based on an example from XPP Studio (Peter Ashwin and Bard Ermentrout, GPL-3.0). Every number below is computed by XPPAUT 8.1 and its AUTO through pyauto-xpp.*

```python
import plotly.graph_objects as go
from pyxpp.builder import ModelBuilder
from xpptut import Explorer, load, orbit_diagram, show_figure
```

## The model

Written with `ModelBuilder`, statement by statement as in an `.ode` file. Edit it and rerun the cells below.

```python
mb = (ModelBuilder("delayed_logistic")
      .ode("x", "r*x*(1-delay(x,tau))")
      .aux("xlag", "delay(x,tau)")
      .parameter(r=1.8, tau=1)
      .initial_value("x", "0.5+0.2*cos(3*t)")
      .option(total=100, dt=0.02, delay=5))
print(mb.render())
```

## Explore

The sliders, the direction-field menu and the AUTO panel all call XPP. `ex.m` is the pyxpp model, for anything else.

```python
ex = Explorer(mb, kind="series", x="x", y="xlag", window=(0, 3, 0, 3), sliders={"r": (0.5, 3), "tau": (0.2, 2)},
              title="Delayed logistic")
ex
```

## Going further

**The Hopf point**: oscillation amplitude at long times against r (τ = 1).

```python
import numpy as np
R = np.linspace(1.2, 2.2, 21)
amp = [np.ptp(ex.m.run(keep=False, r=r, tau=1, total=600, trans=400)["x"]) for r in R]
fig = go.Figure(go.Scatter(x=R, y=amp, mode="lines+markers"))
fig.add_vline(x=np.pi / 2, line_dash="dot", annotation_text="pi/2")
show_figure(fig, xaxis_title="r (tau = 1)", yaxis_title="amplitude", height=300)
```
