# Delayed logistic map: a Neimark-Sacker bifurcation

A two-dimensional map whose fixed point loses stability at a = 2 through a pair of
complex multipliers crossing the unit circle: an invariant closed curve appears
(Neimark-Sacker).

*Based on an example from XPP Studio (Peter Ashwin and Bard Ermentrout, GPL-3.0). Every number below is computed by XPPAUT 8.1 and its AUTO through pyauto-xpp.*

```python
import plotly.graph_objects as go
from pyxpp.builder import ModelBuilder
from xpptut import Explorer, load, orbit_diagram, show_figure
```

## The model

Written with `ModelBuilder`, statement by statement as in an `.ode` file. Edit it and rerun the cells below.

```python
mb = (ModelBuilder("delayed_logistic_map")
      .map("x", "y")
      .map("y", "a*y*(1-x)")
      .parameter(a=2.1)
      .initial(x=0.6, y=0.6)
      .option(total=4000, meth="discrete"))
print(mb.render())
```

## Explore

The sliders, the direction-field menu and the AUTO panel all call XPP. `ex.m` is the pyxpp model, for anything else.

```python
ex = Explorer(mb, kind="map", x="x", y="y", window=(0.1, 0.95, 0.1, 0.95), sliders={"a": (1.5, 2.27)},
              title="Delayed logistic map")
ex
```

## Going further

**The multipliers of the fixed point** (1 − 1/a, 1 − 1/a) cross |λ| = 1 at a = 2.

```python
import numpy as np
m = load(mb)
for a in (1.9, 2.0, 2.1):
    m.pars["a"] = a
    e = m.equilibrium({"x": 1 - 1 / a, "y": 1 - 1 / a})
    print(f"a = {a}: |multipliers| = {np.round(np.abs(e.eigenvalues), 5)}")
```
