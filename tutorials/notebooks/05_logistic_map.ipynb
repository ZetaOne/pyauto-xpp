# The logistic map: period doubling to chaos

x → r x (1 − x). The fixed point loses stability at r = 3, then period 2, 4, 8, ...
accumulate near r = 3.57, after which most `r` give chaos, with windows of periodic
behaviour.

```python
import plotly.graph_objects as go
from pyxpp.builder import ModelBuilder
from xpptut import Explorer, load, orbit_diagram, show_figure
```

## The model

Written with `ModelBuilder`, statement by statement as in an `.ode` file. Edit it and rerun the cells below.

```python
mb = (ModelBuilder("logistic_map")
      .map("x", "r*x*(1-x)")
      .parameter(r=3.7)
      .initial(x=0.5)
      .option(total=300, meth="discrete"))
print(mb.render())
```

## Explore

The sliders, the direction-field menu and the AUTO panel all call XPP. `ex.m` is the pyxpp model, for anything else.

```python
ex = Explorer(mb, kind="series", x="x", y="x", sliders={"r": (2.5, 4)}, title="Logistic map")
ex
```

## Going further

**The orbit diagram**: for each `r`, the last 150 of 650 iterates.

```python
import numpy as np
P, X = orbit_diagram(mb, "r", np.linspace(2.5, 4, 600), "x", transient=500, keep=150)
show_figure(go.Figure(go.Scattergl(x=P, y=X, mode="markers", marker=dict(size=1, color="black"))),
            xaxis_title="r", yaxis_title="x", height=420)
```

**AUTO on the map**: the fixed point and where it loses stability (multiplier −1).

```python
m = load(mb)
m.run(r=2.8); m.run_last()
d = m.auto.run("r", start="steady", ds=0.01, dsmax=0.02, parmin=2.5, parmax=4, nmax=200)
hb = d.special("HB")[0]
print(f"fixed point unstable at r = {hb.x:.6f} (exact 3); multiplier {hb.eigenvalues[0].real:.4f}")
```
